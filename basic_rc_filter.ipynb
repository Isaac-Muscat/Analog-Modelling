```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.signal as signal
```

```python
# Signal BW approaches 8khz, so put cutoff freq around 10khz
# First order
R = 20
C = 200e-9

# Second Order
R1 = 10
C1 = 100e-9
R2 = R1
C2 = C1
```

```python
sys_1 = signal.TransferFunction([1], [2*np.pi*R*C, 1])
sys_2 = signal.TransferFunction([1], [R1*R2*C1*C2, R1*C1+R1*C2+R2*C2, 1])

# Calculate Bode magnitude and phase data
w1, mag1, phase1 = signal.bode(sys_1)
w2, mag2, phase2 = signal.bode(sys_2)
```

```python
def bode_plot(w, mag, phase):
	# Create a figure with two stacked subplots
	fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 6), sharex=True)

	# 1. Magnitude Subplot
	ax1.semilogx(w, mag, color='blue')
	ax1.set_title('Bode Plot')
	ax1.set_ylabel('Magnitude (dB)')
	ax1.grid(True, which="both", ls="-")

	# 2. Phase Subplot
	ax2.semilogx(w, phase, color='red')
	ax2.set_ylabel('Phase (deg)')
	ax2.set_xlabel('Frequency (rad/sec)')
	ax2.grid(True, which="both", ls="-")

	plt.tight_layout()
	plt.show()
```

```python
bode_plot(w1, mag1, phase1)
bode_plot(w2, mag2, phase2)
```
