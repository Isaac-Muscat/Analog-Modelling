```python
from scipy import signal
import matplotlib.pyplot as plt
import numpy as np
```

```python
a = 0.5
N = [(a - a**-1), 0]      # Numerator
D = [1, -(a+a**-1), 1] # Denominator

# 1. Calculate zeros (z) and poles (p)
z, p, k = signal.tf2zpk(N, D)
```

```python
# Plot unit circle
theta = np.linspace(0, 2*np.pi, 100)
x = np.cos(theta)
y = np.sin(theta)
plt.figure(figsize=(5, 5))
plt.plot(x, y, label='Unit Circle')
plt.gca().set_aspect('equal')

plt.plot(z.real, z.imag, 'go', label='Zeros', markersize=10) # 'o' for zeros
plt.plot(p.real, p.imag, 'rx', label='Poles', markersize=10) # 'x' for poles

plt.axhline(0, color='black', lw=1)
plt.axvline(0, color='black', lw=1)
plt.axis('equal')
plt.title('Pole-Zero Plot of Z-Transform')
plt.xlabel('Real')
plt.ylabel('Imaginary')
plt.legend()
plt.grid(True, which='both')
plt.show()
```
