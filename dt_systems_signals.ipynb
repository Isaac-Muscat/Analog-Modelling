```python
N = 10
k = np.linspace(-N/2, N/2, N+1)

d = delta_func(k)
u = np.heaviside(k, 1)
e = 0.5**k
c = np.cos(np.pi*k/5)
```

```python
fig, axs = plt.subplots(nrows=1, ncols=4, figsize=(8,2))
axs[0].stem(k, d, linefmt='r')
axs[0].set_title("Unit Sample")
axs[1].stem(k, u, linefmt='b')
axs[1].set_title("Unit Step")
axs[2].stem(k, e, linefmt='g')
axs[2].set_title("Real Exp")
axs[3].stem(k, c, linefmt='y')
axs[3].set_title("Sinusoidal")
```

```python
# Chapter 2, Question 9a Experiment
N = 10
k = np.linspace(0, N-1, N)
y = np.array([0, 1/3])
s1 = np.array([])
s = np.array([])

# Reference difference equation
for n in range(2, N):
	y = np.append(y, 5/6*y[n-1] - 1/6*y[n-2] + 1/3)

# Incorrect response
for n in range(0, N):
	s1 = np.append(s1, 4*(1/2)**n - 5*(1/3)**n + 1)

# Incorrect response
for n in range(0, N):
	s = np.append(s, -2*(1/2)**n + 1*(1/3)**n + 1)

fig, axs = plt.subplots(nrows=1, ncols=3, figsize=(8,2))
axs[0].stem(k, y, linefmt='r')
axs[0].set_title("Difference Eq")
axs[1].stem(k, s1, linefmt='b')
axs[1].set_title("Incorrect solution")
axs[2].stem(k, s, linefmt='g')
axs[2].set_title("correct solution")
```

```python
# Chapter 2, Question 9c Experiment
N = 10
k = np.linspace(0, N-1, N)
y = np.array([1, 1])
s1 = np.array([])
s = np.array([])

# Reference difference equation
for n in range(2, N):
	y = np.append(y, 5/6*y[n-1] - 1/6*y[n-2])

# Incorrect response
for n in range(0, N):
	s1 = np.append(s1, 4*(1/2)**n - 5*(1/3)**n + 1)

# Incorrect response
for n in range(0, N):
	s = np.append(s, 4*(1/2)**n - 3*(1/3)**n)

fig, axs = plt.subplots(nrows=1, ncols=3, figsize=(8,2))
axs[0].stem(k, y, linefmt='r')
axs[0].set_title("Difference Eq")
axs[1].stem(k, s1, linefmt='b')
axs[1].set_title("Incorrect solution")
axs[2].stem(k, s, linefmt='g')
axs[2].set_title("correct solution")
```
